```python
students = {
    "Name":["Arjun", "Bhavana", "Chirag", "Deepika", "Eshwar", "Farhan", 
        "Gautami", "Hari", "Isha", "Jayesh", "Kavya", "Lokesh", 
        "Manoj", "Nisha", "Omkar", "Pooja"],
    "Subject":["Mathematics", "Physics", "Mathematics", "Mathematics", "Mathematics", "Physics", 
        "Mathematics", "Chemistry", "Chemistry", "Physics", "Physics", "Mathematics", 
        "Mathematics", "Chemistry", "Mathematics", "Chemistry"],
    "Marks":[51, 64, 93, 87, 51, 64, 85, 64, 50, 71, 63, 55, 72, 52, 74, 90],
    "Attendance":[82, 73, 99, 92, 70, 97, 77, 93, 75, 82, 86, 89, 87, 94, 70, 88]
}
df = pd.DataFrame(students)
print(df)
```

```python
math = df.groupby("Subject")
print(math["Marks"].mean())
```

```python
print("Attendance<80%")
print(df[df["Attendance"]<80])
```

```python
df["Status"]= np.where(df["Marks"]>60, "Pass", "Fail")
print(df)
```

```python
top5 = df.nlargest(5, "Marks")
print(top5)
```

```python
df_atten=df.sort_values(by="Attendance")
plt.plot(df_atten["Attendance"],df_atten["Marks"],color = "red",marker="o")
plt.show()
```
