```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df2015 = pd.read_csv("2015.csv")
df2016 = pd.read_csv("2016.csv")
df2017 = pd.read_csv("2017.csv")
df2018 = pd.read_csv("2018.csv")
df2019 = pd.read_csv("2019.csv")
```

```python
print("Happiest 2015")
print(df2015[df2015["Happiness Rank"]==df2015["Happiness Rank"].min()])
print("Least Happy 2015")
print(df2015[df2015["Happiness Rank"]==df2015["Happiness Rank"].max()])
print("Happiest 2019")
print(df2019[df2019["Overall rank"]==df2019["Overall rank"].min()])
print("Least Happy 2019")
print(df2019[df2019["Overall rank"]==df2019["Overall rank"].max()])
```

```python
scores2015 = df2015[["Country", "Happiness Score"]].rename(columns={"Happiness Score": "Score"})
scores2016 = df2016[["Country", "Happiness Score"]].rename(columns={"Happiness Score": "Score"})
scores2017 = df2017[["Country", "Happiness.Score"]].rename(columns={"Happiness.Score": "Score"})
scores2018 = df2018[["Country or region", "Score"]].rename(columns={"Country or region": "Country"})
scores2019 = df2019[["Country or region", "Score"]].rename(columns={"Country or region": "Country"})

allyears = pd.concat([scores2015, scores2016, scores2017, scores2018, scores2019])
scores = allyears.groupby("Country")["Score"].mean().sort_values(ascending=False)
print(scores.head(1))
```

```python
regions= dict(zip(df2015["Country"],df2015["Region"]))
    
regions2015 = df2015.head(10)["Country"].map(regions)
regions2016 = df2016.head(10)["Country"].map(regions)
regions2017 = df2017.head(10)["Country"].map(regions)
regions2018 = df2018.head(10)["Country or region"].map(regions)
regions2019 = df2019.head(10)["Country or region"].map(regions)
total = pd.concat([regions2015, regions2016, regions2017, regions2018, regions2019])
print(total.value_counts())
```

```python
scores2015 = df2015[["Country", "Happiness Score"]].rename(columns={"Happiness Score": "Score"})
scores2016 = df2016[["Country", "Happiness Score"]].rename(columns={"Happiness Score": "Score"})
scores2017 = df2017[["Country", "Happiness.Score"]].rename(columns={"Happiness.Score": "Score"})
scores2018 = df2018[["Country or region", "Score"]].rename(columns={"Country or region": "Country"})
scores2019 = df2019[["Country or region", "Score"]].rename(columns={"Country or region": "Country"})
allyears = pd.concat([scores2015, scores2016, scores2017, scores2018, scores2019],)
allyears["Region"] = allyears["Country"].map(regions)
allyearsfilter = allyears[allyears["Region"]=="Latin America and Caribbean"]
scores = allyearsfilter.groupby("Country")["Score"].mean().sort_values(ascending=False)
print(scores.head(1))
```
