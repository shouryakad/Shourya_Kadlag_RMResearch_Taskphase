```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns
df = pd.read_csv("weatherAUS.csv")
```

```python
df.describe()
```

```python
df.info()
```

```python
#Handling missing data

threshold = len(df) * 0.60
df1 = df.dropna(thresh=threshold, axis=1)
print(df1.shape)

df2 = df1.copy()
df2.dropna(inplace=True)
df2.info()
```

```python
#Encoding the wind directions

df3 = df2.copy()
df3 = pd.get_dummies(df2, columns=['WindGustDir','WindDir9am','WindDir3pm'], dtype=float)
df3
```

```python
#Label Encoding the boolean values

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df4 = df3.copy()
df4['RainToday'] = le.fit_transform(df4['RainToday'])
df4['RainTomorrow'] = le.fit_transform(df4['RainTomorrow'])
df4.loc[0:10,'RainTomorrow']
```

```python
cities = df4.groupby("Location", as_index='False')["MaxTemp"].mean()
plt.figure(figsize=(12, 6))
cities.plot(kind='bar', color='pink')
plt.xlabel('Location', fontsize=12)
plt.ylabel('Average Max Temp', fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#Each city will have different outliers so iqr encoding must be used on one city at a time
grouped_cities = df4.groupby('Location')
df5 = grouped_cities.get_group('Albury').copy()
df5.info()
```

```python
plt.figure(figsize=(10, 5))
plt.hist(df5['MaxTemp'].dropna(), bins=30, color='pink', edgecolor='black')
plt.xlabel('MaxTemp (°C)')
plt.ylabel('Number of days')
plt.show()

plt.figure(figsize=(10, 5))
plt.hist(df5['Rainfall'].dropna(), bins=30, color='pink', edgecolor='black')
plt.xlabel('MaxTemp (°C)')
plt.ylabel('Number of days')
plt.show()
```

```python
#IQR encoding is only required on numerical columns
column_indices = list(range(2, 15))

target_df = df5.iloc[:, column_indices]

q1 = target_df.quantile(0.25)#
q3 = target_df.quantile(0.75)
iqr = q3 - q1

lower_bounds = q1 - 1.5 * iqr
upper_bounds = q3 + 1.5 * iqr

df5.iloc[:, column_indices] = target_df.clip(lower_bounds, upper_bounds, axis=1)
df5.info()
```

```python
#Scaling using normalization
from sklearn.preprocessing import MinMaxScaler

num_cols = ['MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed',
            'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am', 'Humidity3pm',
            'Pressure9am', 'Pressure3pm', 'Cloud9am', 'Temp9am', 'Temp3pm']

scaler = MinMaxScaler()
df6 = df5.copy()
df6[num_cols] = scaler.fit_transform(df5[num_cols])

df6
```

```python
from sklearn.model_selection import train_test_split

X = df6.drop(columns=['RainTomorrow', 'Date', 'Location'])
y = df6['RainTomorrow']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)#

print(X_train.shape, X_test.shape)
```
